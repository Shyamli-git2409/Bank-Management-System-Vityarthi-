```python
bank=[['Aarav',200000,1234567], ['Tigmansh',200000,3456789]]
```

```python
ac=int(input('Enter your account number'))
for i in bank:
    if i[2]==ac:
        acc_selected=i
        index=bank.index(acc_selected)
        print('Your bank details are as follows:',i)
print('''Select your service
1) Deposit
2) Withdraw
3) Check balance
4) Create new account''')
n=int(input('Select your service'))
if n==1:
    deposit_amount=int(input('Enter the amount you want to deposit'))
    bank[index][1]=bank[index][1]+deposit_amount
    print('Amount deposited succesfully!')
elif n==2:
    Withdrawl_amount=int(input('Enter the amount you want to withdraw'))
    if Withdrawl_amount>bank[index][1]:
        print('unsufficient balance')
    else:
        bank[index][1]=bank[index][1]-Withdrawl_amount
        print('Amount withdrawn succesfully!')
elif n==3:
    print('Your account balance is', bank[index][1])
elif n==4:
    n1=input('Enter your name')
    n2=int(input('Enter the amount you want to deposit'))
    import random
    acc_numbers=[]
    for i in bank:
        acc_numbers.append(bank[i][2])
    while True:
        n=random.randrange(1000000,9999999)
        if n not in acc_numbers:
            break
    n3=n
    bank.append([n1,n2,n3])
    print('Account created succesfully!')
    print('Your account number is', n3)
else:
    print('Invalid Choice')
```
